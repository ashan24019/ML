# 1 drop_duplicated()

```python
# drop duplicates in 2 or more columns
df_laptops.drop_duplicates(['Company'])[["Company", "Price_euros"]]
```

```python
# sort dataframe ascending by company and price
# the cheapest first and most expensive last
df_laptops = df_laptops.sort_values(["Company", "Price_euros"])
```

```python
# cheapest: keep='first'
df_laptops.drop_duplicates(["Company"], keep='first')[["Company", "Price_euros"]]
```

```python
# most expensive: keep='last'
df_laptops.drop_duplicates(["Company"], keep='last', inplace=True, ignore_index=True)
```

```python
df_laptops[["Company", "Price_euros"]]
```

##### Task: Find the biggest and smallest screen size in laptops of each company using sort_values() and duplicated() methods (keep "first", "last" and False)

```python
df_laptops = pd.read_csv('../Data Filtering/laptop_price.csv', encoding='latin1')
df_laptops
```

```python
df_laptops_sorted = df_laptops.sort_values(["Company", "Inches"])
```

```python
# smallest

df_laptops_sorted.drop_duplicates("Company", keep="first")[["Company", "Inches"]]
```

```python
# biggest

df_laptops_sorted.drop_duplicates("Company", keep="last")[["Company", "Inches"]]
```
