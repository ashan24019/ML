```python
import pandas as pd
from pandas.core.algorithms import duplicated
```

```python
df_laptops = pd.read_csv('laptop_price.csv', encoding='latin1')
```

```python
df_laptops.head(5)
```

# 1 duplicate()

## 1.1 Example 1

```python
# find duplicates in 1 column/series

df_laptops.duplicated('laptop_ID')
```

```python
# show elements in the dataframe with duplicates in the column "laptop_ID"

df_laptops[df_laptops.duplicated('laptop_ID')]
```

```python
# duplicates in 2 or more columns

df_laptops.duplicated(['TypeName', 'Product', 'Inches'])
```

```python
duplicated = df_laptops.duplicated(['TypeName', 'Product', 'Inches'])
```

```python
# show all the values duplicated (except for the first - keep='first')

df_laptops[duplicated].sort_values(['Product', "TypeName"])
```
