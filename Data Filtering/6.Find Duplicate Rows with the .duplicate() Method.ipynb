# 1 duplicate()

## 1.1 Example 1

```python
# find duplicates in 1 column/series

df_laptops.duplicated('laptop_ID')
```

```python
# show elements in the dataframe with duplicates in the column "laptop_ID"

df_laptops[df_laptops.duplicated('laptop_ID')]
```

```python
# duplicates in 2 or more columns

df_laptops.duplicated(['TypeName', 'Product', 'Inches'])
```

```python
duplicated = df_laptops.duplicated(['TypeName', 'Product', 'Inches'])
df_laptops[duplicated].head(5)
```

```python
# show all the values duplicated (except for the first - keep='first')

df_laptops[duplicated].sort_values(['Product', "TypeName"])
```

## 1.2 Example 2

##### Task: Find the cheapest and most expensive laptop of each company using sort_values() and duplicated() methods (keep= "first" 'last" and False)

```python
# sort the dataframe ascending by company and price
# the cheapest first and most expensive last

df_laptops = df_laptops.sort_values(['Company', 'Price_euros'], ascending=[True, True])

df_laptops
```

```python
# check all values categories in "Company" column
df_laptops.value_counts("Company")
```

```python
# duplicated values in the "Company" column (default -> keep='first')

duplicated_first = df_laptops.duplicated("Company", keep='first')
```

```python
# show dataframe with duplicated values in the "Company" column
df_laptops[duplicated_first]
```

```python
# keep='first'(cheapest laptop per company)
# show dataframe with non duplicated values in "Company" column

df_laptops[~duplicated_first][['Company', 'Price_euros']]
```

```python
# check all value categories
df_laptops[~duplicated_first].value_counts("Company")
```

```python
# keep = "last"
duplicated_last = df_laptops.duplicated("Company", keep='last')
```

```python
# keep='last'(most expensive laptop per company)
# show dataframe with non duplicated values in "Company" column

df_laptops[~duplicated_last][['Company', 'Price_euros']]
```

```python
# keep= "False"
duplicated_false = df_laptops.duplicated("Company", keep=False)
```

```python
# keep= "False"
# show dataframe with non duplicated values in "Company" column
df_laptops[~duplicated_false][['Company', 'Price_euros']]
```

#### every company has at least two laptops. that's why the result is an empty dataframe
